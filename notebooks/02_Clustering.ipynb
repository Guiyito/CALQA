# Notebook 2 — Clustering de Trayectorias FSM

**Inputs:** `outputs/trajectories_summary.csv`  
**Outputs:** `outputs/clusters_k3.csv`, `outputs/clusters_k4.csv`

```python
import os, numpy as np, random, pandas as pd
from sklearn.preprocessing import normalize
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
SEED = 42
np.random.seed(SEED)
random.seed(SEED)
BASE_PATH = '.'
OUTPUTS = os.path.join(BASE_PATH, 'outputs')
print(f'OUTPUTS: {OUTPUTS} | SEED: {SEED}')
```

```python
traj_summary = pd.read_csv(f"{OUTPUTS}/trajectories_summary.csv")
print(f'Shape: {traj_summary.shape}')
traj_summary.head()
```

## Vectorización L2

```python
X = traj_summary.select_dtypes(include=[float, int])
X_norm = normalize(X)
print(f'Shape matriz clustering: {X_norm.shape}')
X.head()
```

## Clustering K=3

```python
k3 = KMeans(n_clusters=3, random_state=SEED, n_init=50)
labels_k3 = k3.fit_predict(X_norm)
sil_k3 = silhouette_score(X_norm, labels_k3)
clusters_k3 = traj_summary.copy()
clusters_k3['cluster'] = labels_k3
clusters_k3.to_csv(f"{OUTPUTS}/clusters_k3.csv", index=False)
print(f'Silhouette K=3: {sil_k3:.4f}')
clusters_k3.groupby('cluster').mean(numeric_only=True).round(3)
```

## Clustering K=4

```python
k4 = KMeans(n_clusters=4, random_state=SEED, n_init=50)
labels_k4 = k4.fit_predict(X_norm)
sil_k4 = silhouette_score(X_norm, labels_k4)
clusters_k4 = traj_summary.copy()
clusters_k4['cluster'] = labels_k4
clusters_k4.to_csv(f"{OUTPUTS}/clusters_k4.csv", index=False)
print(f'Silhouette K=3: {sil_k3:.4f}')
print(f'Silhouette K=4: {sil_k4:.4f}  <- seleccionado')
print('Alumnos por cluster K=4:')
for c, grp in clusters_k4.groupby('cluster'):
    print(f'  Cluster {c}: {list(grp["email"].values)}')
clusters_k4.groupby('cluster').mean(numeric_only=True).round(3)
```
